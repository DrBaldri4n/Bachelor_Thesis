```python
import pandas as pd

df = pd.read_parquet('cleaned_owndata.parquet')

df.head(-1)
```

```python
# Lade deinen bereinigten DataFrame (z.B. df = pd.read_parquet('dein_bereinigter_datensatz.parquet'))
block_size = len(df) // 10
blocks = [df.iloc[i*block_size:(i+1)*block_size] for i in range(9)]
blocks.append(df.iloc[9*block_size:])  # Letzter Block nimmt den Rest
```

```python
label_counts = [block['Label'].value_counts() for block in blocks]
for i, counts in enumerate(label_counts):
    print(f'Block {i+1}:', dict(counts))
```

```python
# Beispiel: Blöcke 2, 5, 8 für Test, Rest für Training
test_blocks = [blocks[1], blocks[4], blocks[8]]
train_blocks = [block for i, block in enumerate(blocks) if i not in [1, 4, 8]]
```

```python
# Test- und Trainingsdaten zusammenfügen
test_df = pd.concat(test_blocks)
train_df = pd.concat(train_blocks)

# Reihenfolge beibehalten (optional: Index zurücksetzen)
test_df = test_df.reset_index(drop=True)
train_df = train_df.reset_index(drop=True)

# Speichern
test_df.to_parquet('test_block_1.parquet')
train_df.to_parquet('train_block_2.parquet')
```

```python
# Parquet-Dateien laden
train_df = pd.read_parquet('train_block_2.parquet')
test_df = pd.read_parquet('test_block_1.parquet')

# Prozentuale Verteilung der Labels berechnen
def label_percentages(df):
    return (df['Label'].value_counts(normalize=True) * 100).round(2)

print("Train-Set:")
print(label_percentages(train_df))

print("\nTest-Set:")
print(label_percentages(test_df))
```
