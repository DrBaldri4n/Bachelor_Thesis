```python
import pandas as pd

df = pd.read_csv('cleaned_owndata.csv')
df.to_parquet('cleaned_owndata.parquet')

df.info()
```

```python
top20_features = [
    "syscalls_sys_exit_recvfrom",
    "bus_access_not_shared",
    "tcp_tcp_retransmit_skb",
    "jbd2_jbd2_checkpoint_stats",
    "l2d_cache_wb",
    "jbd2_jbd2_shrink_count",
    "qdisc_qdisc_dequeue",
    "syscalls_sys_enter_accept4",
    "syscalls_sys_enter_accept",
    "node-loads",
    "bus-cycles",
    "iTLB-load-misses",
    "dTLB-load-misses",
    "node-stores",
    "l1d_cache_wb_clean",
    "syscalls_sys_enter_fchmod",
    "dp_spec",
    "l2d_cache_refill_rd",
    "br_mis_pred",
    "mem_access"
]
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

selected_features = top20_features[:20]
df_subset = df.iloc[3000:]

for feat in selected_features:
    plt.figure(figsize=(14, 6))
    plt.plot(df_subset.index, df_subset[feat], label=feat)
    
    ymin, ymax = plt.ylim()
    start_idx = 0
    current_label = df_subset['Label'].iloc[0]
    current_attack = df_subset['attack'].iloc[0]

    for i in range(1, len(df_subset)):
        label_changed = df_subset['Label'].iloc[i] != current_label
        attack_changed = df_subset['attack'].iloc[i] != current_attack
        if label_changed or attack_changed:
            color = 'green' if current_label == 'benign' else 'red'
            plt.axvspan(df_subset.index[start_idx], df_subset.index[i-1], color=color, alpha=0.1)
            mid_pos = (df_subset.index[start_idx] + df_subset.index[i-1]) / 2
            plt.text(
                mid_pos,
                ymax - (ymax - ymin) * 0.05,
                current_attack,
                ha='center',
                va='top',
                fontsize=9,
                color='black',
                alpha=0.7,
                rotation=90
            )
            start_idx = i
            current_label = df_subset['Label'].iloc[i]
            current_attack = df_subset['attack'].iloc[i]

    # Letztes Segment einfärben und beschriften
    color = 'green' if current_label == 'benign' else 'red'
    plt.axvspan(df_subset.index[start_idx], df_subset.index[-1], color=color, alpha=0.1)
    mid_pos = (df_subset.index[start_idx] + df_subset.index[-1]) / 2
    plt.text(
        mid_pos,
        ymax - (ymax - ymin) * 0.05,
        current_attack,
        ha='center',
        va='top',
        fontsize=9,
        color='black',
        alpha=0.7,
        rotation=90
    )

    plt.title(f'Verlauf von {feat} (ab Zeile 3000)')
    plt.xlabel('Zeilenindex')
    plt.ylabel(feat)
    handles = [
        plt.Line2D([0], [0], color='green', lw=4, alpha=0.1, label='benign'),
        plt.Line2D([0], [0], color='red', lw=4, alpha=0.1, label='attack')
    ]
    plt.legend(handles=handles)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd

# Lade deinen bereinigten DataFrame (z.B. df = pd.read_parquet('dein_bereinigter_datensatz.parquet'))
block_size = len(df) // 10
blocks = [df.iloc[i*block_size:(i+1)*block_size] for i in range(9)]
blocks.append(df.iloc[9*block_size:])  # Letzter Block nimmt den Rest
```

```python
label_counts = [block['Label'].value_counts() for block in blocks]
for i, counts in enumerate(label_counts):
    print(f'Block {i+1}:', dict(counts))
```

```python
# Beispiel: Blöcke 2, 5, 8 für Test, Rest für Training
test_blocks = [blocks[1], blocks[4], blocks[7]]
train_blocks = [block for i, block in enumerate(blocks) if i not in [1, 4, 7]]
```

```python
# Test- und Trainingsdaten zusammenfügen
test_df = pd.concat(test_blocks)
train_df = pd.concat(train_blocks)

# Reihenfolge beibehalten (optional: Index zurücksetzen)
test_df = test_df.reset_index(drop=True)
train_df = train_df.reset_index(drop=True)

# Speichern
test_df.to_parquet('test_owndata_6000_without_feature_selection.parquet')
train_df.to_parquet('train_owndata_6000_without_feature_selection.parquet')
```

```python
# Parquet-Dateien laden
train_df = pd.read_parquet('test_owndata_6000_without_feature_selection.parquet')
test_df = pd.read_parquet('train_owndata_6000_without_feature_selection.parquet')

# Prozentuale Verteilung der Labels berechnen
def label_percentages(df):
    return (df['Label'].value_counts(normalize=True) * 100).round(2)

print("Train-Set:")
print(label_percentages(train_df))

print("\nTest-Set:")
print(label_percentages(test_df))
```
